```python
import numpy as np

consumption = np.array([
    18.2, 19.1, 17.8, 20.4,
    21.0, 19.7, 999.0, -3.0,
    22.5, 31.8, 18.9, 20.1
])
```

## Detect invalid readings
The company has established that valid consumption readings must be greater than zero and less than 100 kWh.
1. Create a Boolean mask identifying valid readings.
2. Use the mask to extract valid readings.
3. Extract invalid readings separately.
4. Count the number of valid and invalid readings.

```python
boolean_mask=(consumption>0)&(consumption<100)
print("Boolean values:",boolean_mask)
valid_readings=consumption[boolean_mask]
print("Valid Readings:",valid_readings)
invalid_readings=consumption[~boolean_mask]
print("Invalid readings:",invalid_readings)
print("Number of valid readings:",sum(boolean_mask))
print("Number of invalid readings:",sum(~boolean_mask))
```

## Clean the dataset
Create a cleaned version of the original array.

**Requirements**:
- Do not modify the original array.
- Replace invalid readings with np.nan.
- Use NumPy functions to calculate the mean, median, minimum, maximum and standard deviation while ignoring missing values.
  
**Expected output**: A cleaned array and a statistical summary.

```python
#Created a another copy for consumption
cleaned_consumption=consumption.astype(float).copy()

invalid_mask=(cleaned_consumption<0)|(cleaned_consumption>100)
cleaned_consumption[invalid_mask]=np.nan
print(cleaned_consumption)

#mean
print("Mean:",np.nanmean(cleaned_consumption))
#Median
print("Median:",np.nanmedian(cleaned_consumption))
#Minimum Value
print("Minimum Value:",np.nanmin(cleaned_consumption))
#Maximum Value
print("Maximum Value:",np.nanmax(cleaned_consumption))
#Standard Deviation
print("Standard Deviation:",round(np.nanstd(cleaned_consumption),2))
```

## Identify unusual consumption
The company wants to investigate unusually high or low energy consumption.

**Use this rule**:
- High consumption: above the mean by more than one standard deviation.
- Low consumption: below the mean by more than one standard deviation.
  
**Your tasks**:
1. Calculate the thresholds.
2. Identify high-consumption readings.
3. Identify low-consumption readings.
4. Display their original day numbers.
   
**Expected output**: A list of unusual readings and the days on which they occurred.

**Important**: Only consider valid readings when identifying unusual consumption.

```python
mean=np.nanmean(cleaned_consumption)
std=round(np.nanstd(cleaned_consumption),2)

days=np.arange(1,len(cleaned_consumption)+1)
print(days)

valid_days=days[boolean_mask]
#Thresholds
High_threshold=mean+std
print("High_threshold:",High_threshold)
Low_threshold=mean-std
print("Low_threshold:",Low_threshold)

#Identifying highthreshold values 
highthreshold_boolean=(valid_readings>High_threshold)
high_threshold_values=valid_readings[highthreshold_boolean]
high_threshold_days=valid_days[highthreshold_boolean]
print("High threshold values :",high_threshold_values)
print("High threshold days:",high_threshold_days)
#Identifying low threshold values
low_threshold_booleans=(valid_readings<Low_threshold)
low_threshold_values=valid_readings[low_threshold_booleans]
low_threshold_days=valid_days[low_threshold_booleans]
print("Low Threshold values:",low_threshold_values)
print("Low threshold Days",low_threshold_days)

```

## Calculate percentage deviation
The company wants to understand how far each valid day's consumption differs from average consumption.

**Calculate the percentage deviation using**:
\[
\text{Percentage deviation} =
\frac{x-\bar{x}}{\bar{x}}\times100
\]

**Where**:
- \(x\) = individual consumption reading.
- \(\bar{x}\) = average valid consumption.
**Requirements**:
- Use vectorized NumPy operations.
- Round the results to two decimal places.
- Preserve the relationship between each reading and its original day.

**Expected output**: A collection of day numbers, consumption readings and percentage deviations.

```python
percentage_deviation=((valid_readings-mean)/mean)*100
percentage_deviation=np.round(percentage_deviation,2)
print("Valid readings:",valid_readings)
print("Valid days:",valid_days)
print("Percentage deviations:",percentage_deviation)
```

## Build a reusable function
Create a function called analyze_energy().

**It should**:
1. Accept a NumPy array as input.
2. Identify invalid readings.
3. Calculate the mean of valid readings.
4. Identify unusually high readings.
5. Return a dictionary containing:
   - valid_count
   - invalid_count
   - mean_consumption
   - high_consumption_days
   
Test your function using the supplied dataset.

Expected output: A dictionary containing the requested results.

```python
import numpy as np
def analyze(x):
    boolean_mask=(x>0)&(x<100)
    valid_readings=x[boolean_mask]
    valid_count=np.sum(boolean_mask)
    invalid_readings=x[~boolean_mask]
    invalid_count=np.sum(~boolean_mask)
    mean_consumption=np.mean(valid_readings)
    std_consumption=np.std(valid_readings)
    high_threshold=mean_consumption+std_consumption
    days=np.arange(1,len(x)+1)
    valid_days=days[boolean_mask]
    high_threshold_boolean=(valid_readings>high_threshold)
    high_threshold_values=valid_readings[high_threshold_boolean]
    high_threshold_days=valid_days[high_threshold_boolean]

    return {"Valid Count:": valid_count,
           "Invalid Count:" : invalid_count,
           "Mean consumption:" : mean_consumption,
           "High consumption days:" : high_threshold_days}
consumption = np.array([
    18.2, 19.1, 17.8, 20.4,
    21.0, 19.7, 999.0, -3.0,
    22.5, 31.8, 18.9, 20.1
])
analyze(consumption)
    
    
```

## Insights and Interpretation

### 1. Data Quality
Out of the 12 recorded electricity consumption readings, 10 were valid while 2 were invalid. The invalid readings were 999 kWh and -3 kWh. The value of 999 kWh was unusually high, while -3 kWh was negative, which is not valid for electricity consumption under the company's rules. These readings were excluded from the statistical analysis.

### 2. Typical Electricity Consumption
The mean electricity consumption was 20.95 kWh, while the median was 19.9 kWh. The mean being slightly higher than the median suggests that the unusually high valid reading of 31.8 kWh increased the average consumption.

### 3. Unusual Consumption
Day 10 recorded the highest valid electricity consumption at 31.8 kWh. It exceeded the high-consumption threshold of 24.79 kWh, making it an unusual reading that should be investigated further. No valid readings were below the low-consumption threshold of 17.11 kWh.

### 4. Percentage Deviation
Day 10's consumption was approximately 51.79% above the average consumption. Day 9's consumption was 7.40% above average, while Day 3 was 15.04% below average. This shows that consumption varied across the recorded days, with Day 10 having the most significant increase.

### 5. Recommendations
- Investigate the high electricity consumption recorded on Day 10 to determine whether it resulted from increased energy usage, equipment activity, or a possible meter issue.
- Check the smart meter that produced the 999 kWh reading to identify the cause of the invalid value.
- Continue monitoring daily electricity consumption to determine whether unusual patterns persist.
- Use the average consumption of 20.95 kWh as a baseline for comparing future readings.
