# Student Attendance Tracker

## Explore the dataset


```python
students = [
    {"name": "Amina", "attended": 18, "total_classes": 20},
    {"name": "Brian", "attended": 14, "total_classes": 20},
    {"name": "Carol", "attended": 20, "total_classes": 20},
    {"name": "David", "attended": 11, "total_classes": 20},
    {"name": "Esther", "attended": 16, "total_classes": 20},
    {"name": "Farah", "attended": 15, "total_classes": 20},
    {"name": "George", "attended": 19, "total_classes": 20},
    {"name": "Hellen", "attended": 13, "total_classes": 20}
]
```

```python
#Display the names of all students
for student in students:
    print(student["name"])
```

```python
#Count the number of students
print("Number of students:",len(students))
```

```python
# Display the attendance record of the first student
print(students[0])
```

## Calculate attendance percentages

```python
for student in students:
    attendance_percentage=(student['attended']/student['total_classes'])*100
    print(f"{student['name']}:{attendance_percentage:0f}%")
    
    
```

```python
print(students[0])
print(type(students[0]))
```

## Classify the students

```python
for student in students:
    attendance_percentage=(student['attended']/student['total_classes'])*100
   
    
    if attendance_percentage>=75:
        print(f"Eligible--{student["name"]}:{attendance_percentage:0f}%")
        
    else:
        print(f"Risk--{student["name"]}:{attendance_percentage:0f}%")
    
```

## Identify at-risk students

```python
at_risk_students=[]

for student in students:
    attendance_percentage=(student["attended"]/student["total_classes"])*100
    if attendance_percentage < 75:
        at_risk_students.append(f"{student["name"]}:{attendance_percentage:0f}%")
        print(f"{student["name"]}:{attendance_percentage:0f}%")

print("The total number of at-risk students:",len(at_risk_students))
```

## Calculate overall attendance

```python
attendance_percentages=[]
for student in students:
    attendance_percentage=(student["attended"]/student['total_classes'])*100
    attendance_percentages.append(attendance_percentage)

avg_attendance=sum(attendance_percentages)/len(attendance_percentages)
print("Average attendance percentage:",avg_attendance,"%")
```

```python
#The student with the highest attendance
highest=max(students, key=lambda student:student['attended']/student['total_classes'])
highest_percentage=(highest['attended']/highest['total_classes'])*100

print(f"Highest attendance--{highest['name']}:{highest_percentage}%")
```

```python
#The student with the lowest attendance
lowest=min(students, key=lambda student:student['attended']/student['total_classes'])
lowest_percentage=(lowest['attended']/lowest['total_classes'])*100

print(f"Lowest attendance--{lowest['name']}:{lowest_percentage:0f}%")
```

## Create a reusable function

```python
def analyze_attendance(x):
  
    attendance_percentages=[]
    eligible_count=[]
    at_risk_students=[]

    for i in x:
        attendance_percentage=(i['attended']/i['total_classes'])*100
        attendance_percentages.append(attendance_percentage)
        if attendance_percentage>=75:
            eligible_count.append(f"{i['name']}:{attendance_percentage:0f}%")
            print(f"Eligible--{i['name']}:{attendance_percentage:0f}%")
        else:
            at_risk_students.append(f"{i['name']}:{attendance_percentage:0f}%")
            print(f"Risk--{i['name']}:{attendance_percentage:0f}%")
        
        
    
    avg_attendance=sum(attendance_percentages)/len(attendance_percentages)
    return {
        "Average Attendance:":avg_attendance,
        "Students who are eligible:":len(eligible_count),
        "Students who are at risk:":len(at_risk_students),
        "At-risk students:":[at_risk_students]
    }
    
analyze_attendance(students)
          
    
    
```

# Insight
The average student attendance is 78.75%, meaning the class meets the university's 75% threshold on average. However, three students are below the required threshold and may need additional academic support
