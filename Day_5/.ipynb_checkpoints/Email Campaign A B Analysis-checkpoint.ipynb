# Email Campaign A/B Analysis

```python
data = [
    {"customer_id": 1, "campaign": "A", "converted": 1},
    {"customer_id": 2, "campaign": "A", "converted": 0},
    {"customer_id": 3, "campaign": "A", "converted": 1},
    {"customer_id": 4, "campaign": "A", "converted": 0},
    {"customer_id": 5, "campaign": "A", "converted": 0},
    {"customer_id": 6, "campaign": "A", "converted": 1},
    {"customer_id": 7, "campaign": "A", "converted": 0},
    {"customer_id": 8, "campaign": "A", "converted": 0},
    {"customer_id": 9, "campaign": "A", "converted": 1},
    {"customer_id": 10, "campaign": "A", "converted": 0},

    {"customer_id": 11, "campaign": "B", "converted": 1},
    {"customer_id": 12, "campaign": "B", "converted": 1},
    {"customer_id": 13, "campaign": "B", "converted": 1},
    {"customer_id": 14, "campaign": "B", "converted": 0},
    {"customer_id": 15, "campaign": "B", "converted": 1},
    {"customer_id": 16, "campaign": "B", "converted": 0},
    {"customer_id": 17, "campaign": "B", "converted": 1},
    {"customer_id": 18, "campaign": "B", "converted": 1},
    {"customer_id": 19, "campaign": "B", "converted": 0},
    {"customer_id": 20, "campaign": "B", "converted": 1},
]
```

## Explore the data

```python
#Count the total customers.
Total=len(data)
print('The total number of customers is :',Total)
```

```python
#Count customers in Campaign A 
Campaign_A=[]
for customers in data:
    if customers['campaign']=='A':
        Campaign_A.append(f"{customers['customer_id']},{customers['converted']}")
CampaignA_customers=len(Campaign_A)
print("The number of customers in Campaign A:",len(Campaign_A))
    
```

```python
#Count customers in Campaign B.
Campaign_B=[]
for customers in data:
    if customers['campaign']=='B':
        Campaign_B.append(f"{customers['customer_id']},{customers['converted']}")
CampaignB_customers=len(Campaign_B)
print('The number of customers in Campaign B:',len(Campaign_B))
```

```python
#Count total conversions
Conversions=[]
for customers in data:
    if customers["converted"]==1:
        Conversions.append(f'{customers['customer_id']},{customers['campaign']}')
Total_conversions=len(Conversions)
print("The total number of conversions :",len(Conversions))
```

```python
#1. Count total non-conversions.
Total_nonconversions=Total-Total_conversions
print("The  total number of non-conversions :",Total_nonconversions)
```

## Calculate conversion rates

```python
#Overall conversion rate.
overall_conversion_rate=round((Total_conversions/Total)*100,2)
print("Overall conversion rate:",overall_conversion_rate,"%")

#Campaign A conversion rate.
CampaignA_conversions=[]
for customers in data:
    if customers['campaign']=="A" and customers['converted']==1:
        CampaignA_conversions.append(f"{customers['customer_id']}")

CampaignA_conversionrate=round((len(CampaignA_conversions)/CampaignA_customers)*100)
print("Campaign A conversion rate:",CampaignA_conversionrate,'%')

#Campaign B conversion rate.
CampaignB_conversions=[]
for customers in data:
    if customers['campaign']=="B" and customers['converted']==1:
        CampaignB_conversions.append(f"{customers['customer_id']}")

CampaignB_conversionrate=round((len(CampaignB_conversions)/CampaignB_customers)*100)
print("Campaign B conversion rate:",CampaignB_conversionrate,'%')
```

## Compare campaign performance

```python
#The difference in conversion rates in percentage points.
Diff_percentagerate=CampaignB_conversionrate-CampaignA_conversionrate
print("The difference in conversion rates in percentage points:",Diff_percentagerate,"%")
#The relative uplift of Campaign B compared with Campaign A.
relative_uplift=(Diff_percentagerate/CampaignA_conversionrate)*100
print("The relative uplift of Campaign B compared with Campaign A:",relative_uplift,"%")
```

## Calculate conditional probabilities

```python
#Probability that the customer converted
prob_conversion=round((len(Conversions)/Total)*100,2)
print("The probability that a customer converted:",prob_conversion,"%")
```

```python
#Probability that the customer received Campaign B
probB_customer=round((CampaignB_customers/Total)*100,2)
print('Probability that the customer received Campaign B:',probB_customer,'%')
```

```python
#Probability that the customer converted given that they received Campaign B
probB_conversion=round((len(CampaignB_conversions)/CampaignB_customers)*100,2)
print("Probability that the customer converted given that they received Campaign B:",probB_conversion,'%')
```

```python
#Probability that a customer who converted had received Campaign B
probB_would_convert=round((len(CampaignB_conversions)/len(Conversions))*100,2)
print("Probability that a customer who converted had received Campaign B",probB_would_convert,'%')
```

## Create a visualization

```python
import matplotlib.pyplot as plt

#Conversion rates
campaigns=['A','B']
conversion_rates=[40,70]

#Create bar chart
plt.bar(campaigns,conversion_rates)

#Add chart title and axis labels
plt.title("Conversion rates by campaign")
plt.xlabel('Campaign')
plt.ylabel('Conversion rates')

#Display values above the bar
for i , rate in enumerate(conversion_rates):
    plt.text(i,rate+1,f'{rate}%',ha='center')

#Display chart
plt.show()
```

# Business interpretation
Campaign B performed better than Campaign A, achieving a conversion rate of 70%, compared with 40% for Campaign A. This represents a 30-percentage-point improvement and a 75% relative uplift.

I would recommend further testing of Campaign B before implementing it across the entire customer base.

However, the dataset contains only 20 customers, with 10 customers per campaign. This is a very small sample and may not accurately represent the company's entire customer population. The company should conduct a larger randomized A/B test and assess statistical significance before making a final decision.
